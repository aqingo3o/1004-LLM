# Introduction
---
model理所當然看不懂自然語言文本(aka我們在看的文字)，他們只看得懂數字，因此我們需要將自然語言文本切成成一個又一個最小單位(文本分割)(電腦認得的"單字"又名`token`)，並對應其ID，電腦才看得懂，舉個例子:  
|token|token ID|
|:---:|:---:|
|dog|1|
|cat|2|
|...|...|

當電腦看到token ID為1，就等同於我們看到"dog"；按照這個方式製作一個`詞彙表`  
現在有兩個分支任務:
1. 對於`訓練文本`:  
    將進行文本分割，接著在用來準備切割訓練資料的dataset利用**滑動視窗法**製作訓練資料對(一對包含一個題目一個答案)，再利用dataloader決定餵給model的資料設定
2. 對於`詞彙表`:  
    給於詞彙表內的每個token(ID)一個專屬的嵌入向量，讓單純的整數ID轉化為具有數學意義的向量，使模型能捕捉詞彙之間的語意關係；訓練model的目的就是在想要知道**詞彙表裡面的語詞**間的語意關係
